## آزمون اولیه کتابخانه‌های عددی و نمودار
یک موج ساختگی فقط برای اطمینان از درست‌بودن اجرای NumPy و Matplotlib رسم می‌شود؛ این نمودار نتیجه تشخیص ژانر نیست.

</div>


```python
import matplotlib.pyplot as plt
t = np.linspace(0, 0.05, 300)
audio_signal = np.sin(2 * np.pi * 440 * t)
plt.figure(figsize=(8, 3))
plt.plot(t, audio_signal)
plt.title("440 Hz sine wave - environment check")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```
